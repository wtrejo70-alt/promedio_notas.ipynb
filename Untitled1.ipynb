```python
# 1. Inicializamos la colección: un diccionario vacío para la agenda de contactos
agenda_contactos = {}

def mostrar_menu():
    print("\n--- 📱 GESTOR DE CONTACTOS ---")
    print("1. Agregar o actualizar contacto")
    print("2. Buscar contacto")
    print("3. Mostrar todos los contactos")
    print("4. Eliminar contacto")
    print("5. Salir")

# Bucle principal del programa
while True:
    mostrar_menu()
    opcion = input("\nSelecciona una opción (1-5): ")

    if opcion == "1":
        # ALMACENAR O ACTUALIZAR DATOS
        nombre = input("Ingresa el nombre del contacto: ").strip().capitalize()
        telefono = input("Ingresa el número telefónico: ").strip()

        # Guardamos en el diccionario usando el nombre como clave y el teléfono como valor
        agenda_contactos[nombre] = telefono
        print(f"¡Contacto '{nombre}' guardado exitosamente!")

    elif opcion == "2":
        # BUSCAR DATOS
        nombre = input("Ingresa el nombre a buscar: ").strip().capitalize()
        if nombre in agenda_contactos:
            print(f"📞 Teléfono de {nombre}: {agenda_contactos[nombre]}")
        else:
            print(f"❌ El contacto '{nombre}' no existe en la agenda.")

    elif opcion == "3":
        # ORGANIZAR Y MOSTRAR INFORMACIÓN
        if not agenda_contactos:
            print("📭 La agenda está vacía.")
        else:
            print("\n📋 LISTA DE CONTACTOS REGISTRADOS:")
            # Ordenamos alfabéticamente por nombre antes de mostrar
            for nombre, telefono in sorted(agenda_contactos.items()):
                print(f"👤 {nombre}: {telefono}")

    elif opcion == "4":
        # ELIMINAR DATOS
        nombre = input("Ingresa el nombre del contacto a eliminar: ").strip().capitalize()
        if nombre in agenda_contactos:
            del agenda_contactos[nombre]
            print(f"🗑️ Contacto '{nombre}' eliminado correctamente.")
        else:
            print(f"❌ El contacto '{nombre}' no se encontró.")

    elif opcion == "5":
        print("👋 ¡Gracias por usar el Gestor de Contactos! Programa finalizado.")
        break

    else:
        print("⚠ Opción no válida. Por favor, intenta de nuevo.")
```
